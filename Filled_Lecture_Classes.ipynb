## Let's make our Spotify Class even more real by having people inputting their username?

We can define that in the `__init__` method or making instance attributes! 

**Instance Attributes** are attributes defined **INSIDE** the initialize function

1. One of the reason reason why we call it the instance attribute is because when you make an instance of the class, you will need to put unique values into it.

2. Another reason is that we can continuously update these some instance attributes it while using our Class

Let's do a series of things!

1. Add The instance Attribute `username`
2. Add The instance Attribute `fav_artist` and assign it to a blank list
3. Add The instance Attribute `fav_playlist`and assign it to a blank list
4. Update `favorite_songs` method to add your favorite artist `fav_artist`
5. Update `favorite_songs` method to add your favorite songs to `fav_playlist`

Caution: 

- Instance Attributes defined as an input paramater `(self)` need to put in every time you make a Class Instance
- Instance Attributes defined inside of the `__init__` method are to be called upon later throughout the entire class

````python
class Spotify:
    platform_name = "Spotify"
    playlists = []
    total_playlists = 0

    def __init__(self, username):
        self.username = username
        self.fav_artist = []
        self.fav_playlist = []
    
    def favorite_songs(self, songs_by_artist):
        lst = []        
        for artist, songs in songs_by_artist.items():
            for song in songs:
                lst.append([song, artist])
            #some conditional that checks if artist is in  the list
            if artist not in self.fav_artist: 
                #append artist name to list
                self.fav_artist.append(artist)
                
        self.fav_playlist = lst

        return lst
````

## But what if we wanted to to create a playlist??

Let's add a Class Method called **`create_playlist`** that returns a **`dictionary`**!

Remember to put the **`self`** paramater inside of out **`create_playlist`** method! 

Make the inputs of `create_playlist` take in a `str` `playlist_name`, a list of your favorite artist, and a list of your favorite songs!

Hint: Use `self.fav_artist` and `self.fav_playlist` as your default parameters

````python
class Spotify:
    platform_name = "Spotify"
    playlists = []
    total_playlists = 0

    def __init__(self, username):
        self.username = username
        self.fav_artist = []
        self.fav_playlist = []
    
    def favorite_songs(self, songs_by_artist):
        lst = []        
        for artist, songs in songs_by_artist.items():
            for song in songs:
                lst.append([song, artist])
            if artist not in self.fav_artist:
                self.fav_artist.append(artist)
        
        self.fav_playlist = lst
        
        return lst

    def create_playlist(self, playlist_name, fav_artist=None, fav_list=None):
        if fav_artist == None:
            fav_artist = self.fav_artist
        if fav_list == None:
            fav_list = self.fav_playlist
    
        lst = []
        #for loop to go through fav_list (embedded list)
        for song, artist in fav_list:
            #conditional to check if the artist is in self.fav_artist
            if artist in self.fav_artist:
                #append a tuple of (songname, artistname) to lst
                lst.append((song,artist))

        #make dictionary
        final_dict = {"name": playlist_name, "songs": lst}
        #append dictionary to self.playlist
        self.playlists.append(final_dict)
        #return dictionary
        return final_dict
````

#### Let's say now that we are older, we will need to show that we are a Graduate Student

Create a new Class Method `introduction_2` that gives us a string statement that we are a Graduate Student

````python
class Grad_Student(CollegeStudent):
    def introduction_2(self):
        #most grad students go to grad school at 22 
        grad_year = self.age - 22 
        if grad_year == 1:
            sequence = 'first'
        elif grad_year == 2:
            sequence = 'second'
        elif grad_year == 3:
            sequence = 'third'
        elif grad_year == 4:
            sequence = 'fourth'
        output = f"Hello, my name is {self.name} I am a {sequence} year Graduate Student at {self.college}" 
        return output
````
